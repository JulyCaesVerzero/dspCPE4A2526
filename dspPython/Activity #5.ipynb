```python
import cv2
import numpy as np
import serial
import time

# Serial communication setup
arduino = serial.Serial('COM1', 115200, timeout=1)
time.sleep(2)

# Start video capture
camera = cv2.VideoCapture(0)

# Define color threshold in HSV (yellow range)
hsv_min = np.array([18, 50, 180])
hsv_max = np.array([35, 255, 255])
margin = 90  # Allowed deviation from center

while True:
    success, img = camera.read()
    if not success:
        continue

    # Convert to HSV
    hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv_img, hsv_min, hsv_max)

    # Get contours from mask
    cnts, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    img_h, img_w = img.shape[:2]
    mid_x = img_w // 2

    # Draw center line
    cv2.line(img, (mid_x, 0), (mid_x, img_h), (0, 0, 255), 2)

    move = ''

    if cnts:
        # Select largest contour
        largest = max(cnts, key=cv2.contourArea)
        x, y, w, h = cv2.boundingRect(largest)
        target_x = x + w // 2
        target_y = y + h // 2

        # Draw visuals
        cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)
        cv2.circle(img, (target_x, target_y), 5, (255, 0, 0), -1)

        # Decide direction
        deviation = target_x - mid_x
        if abs(deviation) > margin:
            move = 'R' if deviation > 0 else 'L'

        # Send to Arduino
        if move:
            arduino.write(move.encode())
            print(f"Sent: {move}")

    # Display both original and mask
    stacked_view = np.hstack((img, cv2.cvtColor(mask, cv2.COLOR_GRAY2BGR)))
    cv2.imshow("Tracking View", stacked_view)

    if cv2.waitKey(1) == 27:
        break

# Cleanup
camera.release()
cv2.destroyAllWindows()
arduino.close()
```
