```python
import cv2
import numpy as np
import serial
import time

# === Serial Initialization ===
def initialize_serial(port='COM1', baud_rate=115200):
    connection = serial.Serial(port, baud_rate, timeout=1)
    time.sleep(2)  # wait for connection to establish
    return connection

# === Frame Processing ===
def detect_colored_object(frame, color_lower, color_upper):
    hsv_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv_frame, color_lower, color_upper)
    contours, _ = cv2.findContours(mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    return mask, contours

# === Determine Movement ===
def calculate_direction(frame_width, object_x, threshold):
    center_screen = frame_width // 2
    deviation = object_x - center_screen

    if abs(deviation) < threshold:
        return None
    return 'R' if deviation > 0 else 'L'

# === Draw Annotations ===
def draw_feedback(frame, x, y, w, h, center_frame_x):
    object_center_x = x + w // 2
    object_center_y = y + h // 2
    cv2.rectangle(frame, (x, y), (x+w, y+h), (50, 205, 50), 2)
    cv2.circle(frame, (object_center_x, object_center_y), 6, (0, 0, 255), -1)
    cv2.line(frame, (center_frame_x, 0), (center_frame_x, frame.shape[0]), (255, 0, 0), 2)

# === Main Routine ===
def start_tracking():
    cam = cv2.VideoCapture(0)
    device = initialize_serial()

    # HSV range for target color (e.g. yellow object)
    target_hsv_min = np.array([18, 50, 180])
    target_hsv_max = np.array([35, 255, 255])
    move_tolerance = 90

    while True:
        ret, video_frame = cam.read()
        if not ret:
            continue

        mask_result, detected_contours = detect_colored_object(video_frame, target_hsv_min, target_hsv_max)
        frame_width = video_frame.shape[1]
        frame_center_x = frame_width // 2

        if detected_contours:
            biggest = max(detected_contours, key=cv2.contourArea)
            x, y, w, h = cv2.boundingRect(biggest)

            draw_feedback(video_frame, x, y, w, h, frame_center_x)

            obj_center_x = x + w // 2
            direction_cmd = calculate_direction(frame_width, obj_center_x, move_tolerance)

            if direction_cmd:
                device.write(direction_cmd.encode())
                print(f"Command dispatched: {direction_cmd}")

        combined_display = np.concatenate(
            (video_frame, cv2.cvtColor(mask_result, cv2.COLOR_GRAY2BGR)), axis=1)
        cv2.imshow("Vision System Output", combined_display)

        if cv2.waitKey(1) == 27:  # ESC key
            break

    cam.release()
    device.close()
    cv2.destroyAllWindows()

# === Entry Point ===
if __name__ == "__main__":
    start_tracking()
```
