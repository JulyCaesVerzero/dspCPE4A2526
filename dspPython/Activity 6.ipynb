```python
# --- MODULATION ---
x_vals = [1, 2, 3, 4, 5, 6]
y_vals = [2, 4, 6, 8, 10, 12]

def modulation(x, y):
    result = []
    for i in range(len(x)):
        result.append(x[i] * y[i])
    return result

mod_output = modulation(x_vals, y_vals)
print("Result of modulation:", mod_output)
```

```python
# --- ADDITION ---
x_vals = [1, 2, 3, 4, 5, 6]
y_vals = [2, 4, 6, 8, 10, 12]

def add_signals(x, y):
    total = []
    for i in range(len(x)):
        total.append(x[i] + y[i])
    return total

sum_output = add_signals(x_vals, y_vals)
print("Result of addition:", sum_output)
```

```python
# --- TIME SHIFT ---
x_vals = [1, 2, 3, 4, 5, 6]
delay = 3

def shift_signal(x, d):
    shifted = [0] * d   # use 0 as filler for delay
    for val in x:
        shifted.append(val)
    return shifted

shifted_output = shift_signal(x_vals, delay)
print("Result of time shift:", shifted_output)
```

```python
# --- SCALAR MULTIPLICATION ---
x_vals = [1, 2, 3, 4, 5, 6]
scale = 4

def scalar_multiply(c, x):
    res = []
    for val in x:
        res.append(c * val)
    return res

scalar_output = scalar_multiply(scale, x_vals)
print("Result of scalar multiplication:", scalar_output)
```
