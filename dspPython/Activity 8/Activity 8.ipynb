```python
import cv2
from ultralytics import YOLO
import numpy as np
import matplotlib.pyplot as plt
```

```python
model = YOLO("best.pt")
```

```python
# Path to your test image (change this to your own image path)
image_path = "mug.png"

# Run detection
results = model(image_path, conf=0.5)  # confidence threshold 50%

# Get the number of detected mugs
num_mugs = len(results[0].boxes) if results[0].boxes is not None else 0

# Print the number of detected mugs
print(f"Number of mugs detected: {num_mugs}")

# For inline display in Jupyter notebook
import cv2
from PIL import Image
import matplotlib.pyplot as plt

# Get annotated image
annotated_img = results[0].plot()

# Add mug count text to the image
text = f"Mugs Detected: {num_mugs}"
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1
font_thickness = 2
text_color = (0, 255, 0)  # Green color
text_position = (10, 30)  # Top-left position

cv2.putText(annotated_img, text, text_position, font, font_scale, text_color, font_thickness)

# Convert BGR to RGB for matplotlib
annotated_img_rgb = cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB)

# Display the image inline in Jupyter
plt.figure(figsize=(10, 8))
plt.imshow(annotated_img_rgb)
plt.axis('off')
plt.title(f"Mug Detection - Found {num_mugs} mugs")
plt.show()

# If you still want the popup window too:
# results[0].show()
```
