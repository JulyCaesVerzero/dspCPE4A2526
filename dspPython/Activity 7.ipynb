```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------
# 1. Create a random sequence
# ------------------------------
n = np.arange(-10, 11)  # index from -10 to 10
np.random.seed(None)    # ensure a new random seed each run
x = np.random.randint(-3, 4, size=len(n))  # random ints between -3 and 3

# ------------------------------
# 2. Time reversal
# ------------------------------
x_rev = x[::-1]  # x[-n]

# ------------------------------
# 3. Even and Odd components
# ------------------------------
x_even = 0.5 * (x + x_rev)
x_odd  = 0.5 * (x - x_rev)

# ------------------------------
# 4. Plot results
# ------------------------------
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.stem(n, x, linefmt='r-', markerfmt='ro', basefmt='k-')
plt.title('Original Sequence x[n]')
plt.xlabel('n'); plt.ylabel('Amplitude')

plt.subplot(2,2,2)
plt.stem(n, x_rev, linefmt='g-', markerfmt='go', basefmt='k-')
plt.title('Time Reverse x[-n]')
plt.xlabel('n'); plt.ylabel('Amplitude')

plt.subplot(2,2,3)
plt.stem(n, x_even, linefmt='m-', markerfmt='mo', basefmt='k-')
plt.title('Even Component')
plt.xlabel('n'); plt.ylabel('Amplitude')

plt.subplot(2,2,4)
plt.stem(n, x_odd, linefmt='c-', markerfmt='co', basefmt='k-')
plt.title('Odd Component')
plt.xlabel('n'); plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```
